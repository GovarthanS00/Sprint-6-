# Year

## What is it?

Year means extracting the **year** from a date.

For example:

`2026-09-01 → 2026`

## Why is it required?

To identify **year-based patterns and trends** in the data.

## When should we use it?

Use it when the **year of an event can affect the prediction or analysis**.

## What does the Python example tell us?

The `.dt.year` operation extracts the **year** from the date column.

## Real-World Example

A company can analyze how many customers made purchases in **2024, 2025, and 2026**.

## Business Example

A retailer can compare **yearly sales** to identify whether revenue is increasing or decreasing.

## AI/ML Example

In a sales prediction model, `purchase_year` can help the model learn **long-term yearly trends**.


```python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "purchase_date": ["2024-03-15", "2025-07-20", "2026-01-10"]
})

df["purchase_date"] = pd.to_datetime(df["purchase_date"])

df["purchase_year"] = df["purchase_date"].dt.year

print(df)
```

# Weekend Indicator

## What is it?

Weekend Indicator identifies whether a date falls on a **Saturday or Sunday**.

* `1` → Weekend
* `0` → Weekday

For example:

`2026-09-05 → Saturday → 1`

## Why is it required?

To help models identify **different behavior between weekdays and weekends**.

## When should we use it?

Use it when customer activity or business operations **change during weekends**.


## What does the Python example tell us?

`dayofweek >= 5` identifies **Saturday and Sunday**, and `.astype(int)` converts the result into `0` or `1`.

## Real-World Example

An e-commerce company can check whether customers place more orders on **weekends**.

## Business Example

A restaurant can use weekend indicators to plan **staffing and inventory**.

## AI/ML Example

A sales prediction model can use `is_weekend` to learn whether **weekend demand is higher than weekday demand**.


```python
## Python Example

import pandas as pd

df = pd.DataFrame({
    "purchase_date": ["2026-09-04", "2026-09-05", "2026-09-06"]
})

df["purchase_date"] = pd.to_datetime(df["purchase_date"])

df["is_weekend"] = (df["purchase_date"].dt.dayofweek >= 5).astype(int)

print(df)
```

# Days Since Event

## What is it?

Days Since Event means calculating **how many days have passed since a specific event**.

For example:

`Purchase Date = 2026-08-25`
`Today = 2026-09-01`

Days since purchase = **7 days**

## Why is it required?

To measure **how recently an event happened**.

## When should we use it?

Use it when the **time elapsed since an event** can affect analysis or prediction.

## What does the Python example tell us?

It calculates the **number of days between today and the purchase date**.

## Real-World Example

A company can identify how many days it has been since a customer **last purchased a product**.

## Business Example

A marketing team can send a reminder to customers who have **not purchased for 30 days**.

## AI/ML Example

A customer churn model can use `days_since_purchase` to identify customers who may be **likely to stop using the service**.


```python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "purchase_date": ["2026-08-25", "2026-08-28", "2026-08-30"]
})

df["purchase_date"] = pd.to_datetime(df["purchase_date"])

today = pd.Timestamp("2026-09-01")

df["days_since_purchase"] = (today - df["purchase_date"]).dt.days

print(df)
```

# Time Difference

## What is it?

Time Difference means calculating the **amount of time between two dates or timestamps**.

For example:

`Start = 2026-09-01 10:00`
`End = 2026-09-01 14:30`

Time difference = **4 hours 30 minutes**

## Why is it required?

To measure **how long an activity or process takes**.

## When should we use it?

Use it when the **duration between two events** is important.

## What does the Python example tell us?

It calculates the **duration between two timestamps** and converts the result into hours.

## Real-World Example

An e-commerce company can calculate how long it takes to **deliver an order**.

## Business Example

A customer support team can calculate the **time taken to resolve a support ticket**.

## AI/ML Example

A delivery prediction model can use `time_difference_hours` to learn **how long deliveries usually take**.


```python

## Python Example
import pandas as pd

df = pd.DataFrame({
    "start_time": ["2026-09-01 10:00", "2026-09-01 09:30"],
    "end_time": ["2026-09-01 14:30", "2026-09-01 12:00"]
})

df["start_time"] = pd.to_datetime(df["start_time"])
df["end_time"] = pd.to_datetime(df["end_time"])

df["time_difference_hours"] = (
    df["end_time"] - df["start_time"]
).dt.total_seconds() / 3600

print(df)
```

# Customer Tenure

## What is it?

Customer Tenure means calculating **how long a customer has been with a company**.

For example:

`Joining Date = 2023-06-01`
`Reference Date = 2026-09-01`

Customer Tenure = **3.25 years approximately**

## Why is it required?

To understand the **length of the customer relationship**.

## When should we use it?

Use it when customer behavior may depend on **how long they have been with the company**.

## What does the Python example tell us?

It calculates the number of **days and approximate years** a customer has been with the company.

## Real-World Example

A telecom company can identify customers who have been using its service for **many years**.

## Business Example

A company can provide **loyalty rewards** to customers with high tenure.

## AI/ML Example

A churn prediction model can use `tenure_days` to learn whether **new customers or long-term customers are more likely to leave**.


```python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "joining_date": ["2023-06-01", "2025-01-15", "2026-01-01"]
})

df["joining_date"] = pd.to_datetime(df["joining_date"])

reference_date = pd.Timestamp("2026-09-01")

df["tenure_days"] = (
    reference_date - df["joining_date"]
).dt.days

df["tenure_years"] = df["tenure_days"] / 365

print(df)

```

# Recency

## What is it?

Recency means measuring **how recently a customer performed an activity**, such as making a purchase.

For example:

`Last Purchase = 2026-08-25`
`Reference Date = 2026-09-01`

Recency = **7 days**

## Why is it required?

To identify **how recently a customer interacted with the business**.

## When should we use it?

Use it when recent customer activity is important for **customer behavior or prediction**.


## What does the Python example tell us?

`recency_days` tells us **how many days have passed since the customer's last purchase**.

**Lower recency = more recent activity.**

**Higher recency = customer has been inactive longer.**

## Real-World Example

An e-commerce company can identify customers who **recently purchased** and customers who have not purchased for a long time.

## Business Example

A marketing team can send special offers to customers whose **recency is high**.

## AI/ML Example

Recency is an important feature in **customer churn prediction and customer segmentation**.

For example:

* `2 days` → highly active customer
* `60 days` → potentially inactive customer


```python
## Python Example

import pandas as pd

df = pd.DataFrame({
    "customer": ["A", "B", "C"],
    "last_purchase": ["2026-08-25", "2026-08-28", "2026-08-30"]
})

df["last_purchase"] = pd.to_datetime(df["last_purchase"])

reference_date = pd.Timestamp("2026-09-01")

df["recency_days"] = (
    reference_date - df["last_purchase"]
).dt.days

print(df)

```

# Frequency

## What is it?

Frequency means measuring **how many times a customer performs an activity** during a specific period.

For example:

A customer made **5 purchases in the last 30 days**.

So:

`Frequency = 5`

## Why is it required?

To identify **how often a customer interacts with the business**.

## When should we use it?

Use it when the number of customer activities can help understand **customer behavior**.

## What does the Python example tell us?

`groupby("customer").size()` counts **how many purchases each customer made**.

For example:

* Customer A → 2 purchases
* Customer B → 3 purchases
* Customer C → 1 purchase

## Real-World Example

An e-commerce company can identify customers who **purchase frequently**.

## Business Example

A company can provide **loyalty rewards** to customers with high purchase frequency.

## AI/ML Example

Frequency can be used as a feature in **customer segmentation and churn prediction**.

For example:

* Frequency = `10` → frequent customer
* Frequency = `1` → occasional customer


```python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "customer": ["A", "A", "B", "B", "B", "C"],
    "purchase_date": [
        "2026-08-01",
        "2026-08-15",
        "2026-08-05",
        "2026-08-10",
        "2026-08-25",
        "2026-08-20"
    ]
})

df["purchase_date"] = pd.to_datetime(df["purchase_date"])

frequency = df.groupby("customer").size().reset_index(name="purchase_frequency")

print(frequency)
```

# Seasonality

## What is it?

Seasonality means a **repeating pattern in data at regular time intervals**.

For example:

* Sales increase every **December**
* Food orders increase every **weekend**
* Electricity usage increases during **summer**

## Why is it required?

To help identify **repeating time-based patterns** that can affect business activity and predictions.

## When should we use it?

Use it when the target variable shows a **regular pattern based on month, quarter, day, or season**.


## What does the Python example tell us?

The `purchase_month` feature helps the model identify **which month the purchase happened**.

If sales repeatedly increase during a particular month, the model can learn that **seasonal pattern**.

## Real-World Example

An e-commerce company may have **higher sales during festival seasons**.

## Business Example

A retailer can increase **inventory before high-demand seasons**.

## AI/ML Example

A sales forecasting model can use date-based features such as `month`, `quarter`, and `day_of_week` to learn **seasonal demand patterns**.


```python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "purchase_date": [
        "2026-01-10",
        "2026-02-10",
        "2026-03-10",
        "2026-12-10"
    ],
    "sales": [100, 120, 130, 300]
})

df["purchase_date"] = pd.to_datetime(df["purchase_date"])

df["purchase_month"] = df["purchase_date"].dt.month

print(df)
```
