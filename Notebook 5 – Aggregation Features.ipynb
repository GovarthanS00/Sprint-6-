# Aggregation Features

## What is it?

**Aggregation features** combine multiple rows into a single summary value.

They help convert detailed data into useful information such as:

* Total customer spending
* Average order value
* Number of purchases
* Maximum transaction amount
* Average product rating


## What does the Python example tell us?

The transaction-level data is converted into **customer-level features**.

For Customer A:

```text
Total Spending      = 600
Average Spending    = 200
Number of Purchases = 3
Maximum Spending    = 300
```

## Real-World Example

An e-commerce company can calculate:

```text
Customer → Total Spending
Customer → Number of Purchases
Customer → Average Order Value
Customer → Maximum Purchase
```

## Business Example

A bank can calculate customer transaction statistics such as:

```text
Total Monthly Spending
Average Transaction Amount
Number of Transactions
Maximum Transaction Amount
```

## AI/ML Use Case

Aggregation features can be used for:

* Customer churn prediction
* Fraud detection
* Credit risk prediction
* Customer segmentation
* Recommendation systems

For example, a churn prediction model can use:

```text
Total_Spending
Number_of_Purchases
Average_Order_Value
Maximum_Transaction
```


````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Customer": ["A", "A", "A", "B", "B"],
    "Amount": [100, 200, 300, 150, 250]
})

customer_features = df.groupby("Customer")["Amount"].agg(
    Total_Spending="sum",
    Average_Spending="mean",
    Number_of_Purchases="count",
    Maximum_Spending="max"
)

print(customer_features)
````

# Count Aggregation

## What is it?

**Count** calculates the number of records in a group.

For example, we can count how many purchases each customer has made.

## What does the Python example tell us?

* Customer A made **3 purchases**
* Customer B made **2 purchases**
* Customer C made **1 purchase**

## Real-World Example

An e-commerce company can count the **number of orders made by each customer**.

```text
Customer A → 3 orders
Customer B → 2 orders
Customer C → 1 order
```

## Business Example

A bank can count the **number of transactions made by each customer** to understand transaction frequency.

## AI/ML Use Case

In customer churn prediction, `Number_of_Purchases` can be used as a feature.

```text
Customer A → Number_of_Purchases = 3
Customer B → Number_of_Purchases = 2
Customer C → Number_of_Purchases = 1
```


````python
## Python Example

import pandas as pd

df = pd.DataFrame({
    "Customer": ["A", "A", "A", "B", "B", "C"],
    "Amount": [100, 200, 300, 150, 250, 400]
})

purchase_count = df.groupby("Customer")["Amount"].count()

print(purchase_count)
````

# Sum Aggregation

## What is it?

**Sum** calculates the total value of a group.

For example, we can calculate the total amount spent by each customer.


## What does the Python example tell us?

* Customer A spent **600**
* Customer B spent **400**
* Customer C spent **400**

## Real-World Example

An e-commerce company can calculate the **total spending of each customer**.

```text
Customer A → ₹600
Customer B → ₹400
Customer C → ₹400
```

## Business Example

A bank can calculate the **total transaction amount** for each customer.

## AI/ML Use Case

In customer churn prediction, `Total_Spending` can be used as a feature.

```text
Customer A → Total_Spending = 600
Customer B → Total_Spending = 400
Customer C → Total_Spending = 400
```


````python
## Python Example

import pandas as pd

df = pd.DataFrame({
    "Customer": ["A", "A", "A", "B", "B", "C"],
    "Amount": [100, 200, 300, 150, 250, 400]
})

total_spending = df.groupby("Customer")["Amount"].sum()

print(total_spending)
````

# Median Aggregation

## What is it?

**Median** is the middle value of a group after sorting the values.

It is useful when data contains very large or very small values that can affect the average.

## What does the Python example tell us?

For Customer A:

```text
100, 200, 1000
```

The middle value is:

```text
200
```

For Customer B:

```text
150, 250, 300
```

The middle value is:

```text
250
```

## Real-World Example

An e-commerce company can calculate the **median order value** for each customer.

## Business Example

A company can use median salary instead of average salary when a few employees have extremely high salaries.

## AI/ML Use Case

In customer behavior prediction, `Median_Order_Value` can be used as a feature when transaction amounts contain extreme values.

```text id="x4v9kp"
Customer A → Median_Order_Value = 200
Customer B → Median_Order_Value = 250
```


````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Customer": ["A", "A", "A", "B", "B", "B"],
    "Amount": [100, 200, 1000, 150, 250, 300]
})

median_spending = df.groupby("Customer")["Amount"].median()

print(median_spending)
````

# Standard Deviation Aggregation

## What is it?

**Standard deviation** measures how much the values in a group vary from their average.

* Low standard deviation → Values are close together
* High standard deviation → Values are more spread out

## What does the Python example tell us?

Customer A:

```text
100, 200, 300
```

The values are more spread out, so the standard deviation is **higher**.

Customer B:

```text
190, 200, 210
```

The values are close together, so the standard deviation is **lower**.

## Real-World Example

An e-commerce company can calculate the **variation in a customer's order amounts**.

## Business Example

A bank can identify customers whose transaction amounts vary significantly.

## AI/ML Use Case

`Transaction_Amount_Std` can be used as a feature for:

* Fraud detection
* Customer behavior analysis
* Risk prediction

```text id="f2k8qn"
Customer A → Higher variation
Customer B → Lower variation
```


````python
## Python Example

import pandas as pd

df = pd.DataFrame({
    "Customer": ["A", "A", "A", "B", "B", "B"],
    "Amount": [100, 200, 300, 190, 200, 210]
})

std_amount = df.groupby("Customer")["Amount"].std()

print(std_amount)
````

# Group-Based Aggregation

## What is it?

**Group-Based Aggregation** calculates summary values separately for each group.

We use `groupby()` to create groups and aggregation functions such as:

* `count()`
* `sum()`
* `mean()`
* `median()`
* `min()`
* `max()`
* `std()`


## What does the Python example tell us?

The data is grouped by **Customer** and multiple features are calculated for each customer.

```text id="f8n2kd"
Customer A
Total Spending       = 300
Average Spending     = 150
Number of Purchases  = 2
Maximum Transaction  = 200
```

```text id="r5q1zx"
Customer B
Total Spending       = 700
Average Spending     = 233.3
Number of Purchases  = 3
Maximum Transaction  = 300
```

## Real-World Example

An e-commerce company can group transactions by customer and calculate:

```text id="m6t4qp"
Total Spending
Average Order Value
Number of Orders
Maximum Order Value
```

## Business Example

A company can group sales by **product category** and calculate total sales and average sales.

## AI/ML Use Case

Group-based aggregation can create features for:

* Customer churn prediction
* Fraud detection
* Customer segmentation
* Sales prediction

```text id="e9c3wl"
Customer → Total Spending
Customer → Average Spending
Customer → Purchase Count
Customer → Maximum Transaction
```


````python
## Python Example

import pandas as pd

df = pd.DataFrame({
    "Customer": ["A", "A", "B", "B", "B"],
    "Amount": [100, 200, 150, 250, 300]
})

customer_summary = df.groupby("Customer")["Amount"].agg(
    Total_Spending="sum",
    Average_Spending="mean",
    Number_of_Purchases="count",
    Maximum_Transaction="max"
)

print(customer_summary)
````


# Customer-Level Aggregation

## What is it?

**Customer-Level Aggregation** creates summary features for each customer using their transaction history.

For example:

* Total customer spending
* Average order value
* Number of purchases
* Maximum transaction amount

## What does the Python example tell us?

Customer A has:

```text id="w6k3rp"
Total Spending       = 800
Average Order Value  = 266.7
Number of Purchases  = 3
Maximum Transaction  = 500
Average Rating       = 4.0
```

These values summarize the customer's transaction history into useful features.

## Real-World Example

An e-commerce company can create a customer profile using:

```text id="j8m2qa"
Total Spending
Number of Purchases
Average Order Value
Maximum Transaction
Average Rating
```

## Business Example

A retail company can identify **high-value customers** by looking at total spending and purchase frequency.

## AI/ML Use Case

Customer-level aggregation features can be used for:

* Customer churn prediction
* Customer segmentation
* Customer lifetime value prediction
* Recommendation systems
* Fraud detection


````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Customer": ["A", "A", "A", "B", "B", "C"],
    "Amount": [100, 200, 500, 150, 250, 400],
    "Rating": [4, 5, 3, 5, 4, 5]
})

customer_features = df.groupby("Customer").agg(
    Total_Spending=("Amount", "sum"),
    Average_Order_Value=("Amount", "mean"),
    Number_of_Purchases=("Amount", "count"),
    Maximum_Transaction=("Amount", "max"),
    Average_Rating=("Rating", "mean")
)

print(customer_features)
````

# Product-Level Aggregation

## What is it?

**Product-Level Aggregation** creates summary features for each product using its transaction or rating history.

For example:

* Total units sold
* Average product rating
* Number of purchases
* Maximum selling price

## What does the Python example tell us?

For the Laptop:

```text
Total Quantity Sold = 6
Average Price       = 50000
Number of Purchases = 3
Maximum Price       = 52000
Average Rating      = 4.33
```

For the Phone:

```text
Total Quantity Sold = 11
Average Price       = 21000
Number of Purchases = 3
Maximum Price       = 22000
Average Rating      = 4.67
```

## Real-World Example

An e-commerce company can calculate product-level features such as:

```text
Total Sales
Average Price
Number of Purchases
Average Rating
Total Quantity Sold
```

## Business Example

A company can identify **best-selling products** by comparing total quantity sold for each product.

## AI/ML Use Case

Product-level aggregation features can be used for:

* Product recommendation
* Demand forecasting
* Sales prediction
* Inventory optimization
* Product ranking


````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Product": ["Laptop", "Laptop", "Laptop", "Phone", "Phone", "Phone"],
    "Quantity": [2, 3, 1, 5, 2, 4],
    "Price": [50000, 52000, 48000, 20000, 21000, 22000],
    "Rating": [4, 5, 4, 5, 4, 5]
})

product_features = df.groupby("Product").agg(
    Total_Quantity_Sold=("Quantity", "sum"),
    Average_Price=("Price", "mean"),
    Number_of_Purchases=("Quantity", "count"),
    Maximum_Price=("Price", "max"),
    Average_Rating=("Rating", "mean")
)

print(product_features)

````
