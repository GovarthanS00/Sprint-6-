# Categorical Variables

## What is it?

Categorical variables are features that contain **categories or labels** instead of numerical values.

**Examples:**

* Gender → Male, Female
* City → Chennai, Bangalore, Mumbai
* Department → HR, IT, Finance

## Why is it required?

Machine Learning models need categorical values to be **converted into a suitable numerical format** before they can use them.

## When should we use it?

Use categorical variables when a feature represents **groups, types, names, or categories**.

## Real-World Example

A person's **city** can be a categorical variable:

```text
Chennai
Bangalore
Mumbai
Delhi
```

These are categories, not numerical measurements.

## Business Example

A company's employee database may contain:

```text
Department
----------
IT
HR
Finance
Sales
Marketing
```

The `Department` column is a categorical variable used to group employees.

## AI/ML Example

In a **customer churn prediction** model, features may include:

```text
Gender       → Male / Female
Contract     → Monthly / Yearly
PaymentMode  → Credit Card / UPI / Cash
```

These categorical features can be converted into numerical values using techniques such as **One-Hot Encoding, Ordinal Encoding, or Target Encoding**.

## What does the Python example tell us?

The `Department` column contains **categories such as IT, HR, and Finance**, so it is a categorical variable.


````python

## Python Example
import pandas as pd

data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Divya"],
    "Department": ["IT", "HR", "IT", "Finance"]
})

print(data)
````

# Nominal Variables

## What is it?

Nominal variables are categorical variables where the categories have **no natural order or ranking**.

**Example:**

```text
City → Chennai, Bangalore, Mumbai
Color → Red, Blue, Green
Department → IT, HR, Finance
```

There is no meaning like one category being higher or lower than another.

## Why is it required?

It helps us identify categorical data where **categories should not be treated as ordered values**.

## When should we use it?

Use nominal variables when categories are **different names or groups without any ranking**.

## Real-World Example

A person's **blood group** is a nominal variable:

```text
A
B
AB
O
```

There is no ranking such as A > B > AB > O.

## Business Example

A company may classify customers by **region**:

```text
South
North
East
West
```

These are different groups, but no region is naturally higher or lower than another.

## AI/ML Example

In a **customer churn prediction** model:

```text
Payment Method
----------------
UPI
Credit Card
Cash
Net Banking
```

These are nominal categories because there is **no meaningful order** between payment methods.

For such data, **One-Hot Encoding** is commonly suitable.

## What does the Python example tell us?

The `Department` column contains different categories (`IT`, `HR`, `Finance`) with **no natural order**, so it is a **nominal variable**.


````python

## Python Example

import pandas as pd

data = pd.DataFrame({
    "Employee": ["Arun", "Priya", "Kumar", "Divya"],
    "Department": ["IT", "HR", "Finance", "IT"]
})

print(data)

````

# Ordinal Variables

## What is it?

Ordinal variables are categorical variables where the categories have a **meaningful order or ranking**.

**Example:**

```text
Satisfaction → Low < Medium < High
Education → School < Bachelor < Master < PhD
```

The order is meaningful, but the **difference between categories is not necessarily equal**.

## Why is it required?

It helps us preserve the **ranking information** in categorical data.

## When should we use it?

Use ordinal variables when categories have a **clear natural order**.


## Real-World Example

A hotel customer may rate their experience as:

```text
Poor < Average < Good < Excellent
```

These categories have a natural order.

## Business Example

A company may classify employee performance as:

```text
Low < Medium < High
```

The ranking can be useful for performance analysis and decision-making.

## AI/ML Example

In a **customer satisfaction prediction** model:

```text
Satisfaction
------------
Low
Medium
High
```

We can use **Ordinal Encoding**:

```text
Low    → 0
Medium → 1
High   → 2
```

This preserves the meaningful order.

## What does the Python example tell us?

The `Satisfaction` column contains `Low`, `Medium`, and `High`, which have a **natural ranking**, so it is an **ordinal variable**.


````python
## Python Example

import pandas as pd

data = pd.DataFrame({
    "Customer": ["Arun", "Priya", "Kumar", "Divya"],
    "Satisfaction": ["Low", "High", "Medium", "High"]
})

print(data)
````

# Binary Variables

## What is it?

Binary variables are categorical variables that have **only two possible categories**.

**Examples:**

```text
Yes / No
True / False
Pass / Fail
Male / Female
```

## Why is it required?

Binary variables are simple to represent using **two numerical values**, usually `0` and `1`.

## When should we use it?

Use binary variables when a feature has **exactly two categories**.


## Real-World Example

A person may have a driving license:

```text
Has License → Yes / No
```

There are only two possible categories.

## Business Example

An online business may record whether a customer purchased a product:

```text
Purchased → Yes / No
```

This can be used to analyze customer purchasing behavior.

## AI/ML Example

In a **customer purchase prediction** model:

```text
Purchased
---------
Yes → 1
No  → 0
```

The binary categorical feature can be converted into `0` and `1` and provided to the ML model.

## What does the Python example tell us?

The `Subscribed` column contains only **two categories: Yes and No**, so it is a **binary variable**.


````python
## Python Example

import pandas as pd

data = pd.DataFrame({
    "Customer": ["Arun", "Priya", "Kumar", "Divya"],
    "Subscribed": ["Yes", "No", "Yes", "No"]
})

print(data)
````

# One-Hot Encoding

## What is it?

One-Hot Encoding converts a categorical variable into **multiple binary columns (0 and 1)**.

For example:

```text
Department → IT, HR, Finance
```

becomes:

```text
Department_IT  Department_HR  Department_Finance
1              0              0
0              1              0
0              0              1
```

## Why is it required?

It converts categorical values into numbers **without creating a false ranking between categories**.

## When should we use it?

Use One-Hot Encoding mainly for **nominal categorical variables** where categories have **no natural order**.

## Real-World Example

A person's **favorite color** can be:

```text
Red
Blue
Green
```

One-Hot Encoding converts these into separate columns:

```text
Color_Red  Color_Blue  Color_Green
```

## Business Example

An e-commerce company may have a `Payment Method` column:

```text
UPI
Credit Card
Cash
Net Banking
```

One-Hot Encoding creates a separate column for each payment method.

## AI/ML Example

In a **customer churn prediction** model:

```text
Contract Type
-------------
Monthly
Yearly
Two-Year
```

One-Hot Encoding converts it into:

```text
Contract_Monthly
Contract_Yearly
Contract_Two-Year
```

This allows the ML model to use the categorical feature without assuming that one contract type is greater than another.

## What does the Python example tell us?

The `Department` categories are converted into **separate 0/1 columns**, so the ML model can use the categorical data without assigning an artificial order such as `IT > HR > Finance`.


````python

## Python Example
import pandas as pd

data = pd.DataFrame({
    "Department": ["IT", "HR", "Finance", "IT"]
})

encoded_data = pd.get_dummies(data, columns=["Department"], dtype=int)

print(encoded_data)

````

# Ordinal Encoding

## What is it?

Ordinal Encoding converts categories into **numbers while preserving their natural order**.

For example:

```text
Low    → 0
Medium → 1
High   → 2
```

## Why is it required?

It converts ordered categorical values into numbers **without losing their ranking**.

## When should we use it?

Use Ordinal Encoding when the categories have a **meaningful natural order**.

## Real-World Example

Customer satisfaction can be:

```text
Low < Medium < High
```

The order is meaningful.

## Business Example

A company can classify customer priority as:

```text
Low < Medium < High
```

This ranking can help the business identify which customers need more attention.

## AI/ML Example

In a **customer satisfaction prediction** model:

```text
Low    → 0
Medium → 1
High   → 2
```

The model receives numerical values while the original ranking is preserved.

## What does the Python example tell us?

The `Satisfaction` categories are converted to `0`, `1`, and `2` according to their **natural order**: `Low < Medium < High`.

**Important:** Do not use Ordinal Encoding for nominal categories such as `IT`, `HR`, and `Finance`, because there is no meaningful order between them.


````python

## Python Example
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

data = pd.DataFrame({
    "Satisfaction": ["Low", "High", "Medium", "High"]
})

encoder = OrdinalEncoder(
    categories=[["Low", "Medium", "High"]]
)

data["Satisfaction_Encoded"] = encoder.fit_transform(
    data[["Satisfaction"]]
).astype(int)

print(data)
````

# Label Encoding

## What is it?

Label Encoding converts each category into a **unique integer value**.

For example:

```text
Red   → 0
Blue  → 1
Green → 2
```

Unlike Ordinal Encoding, these numbers **do not necessarily represent a meaningful order**.

## Why is it required?

It converts categorical labels into **numerical values** that an ML model can process.

## When should we use it?

Use Label Encoding mainly for the **target/output variable (`y`)**, especially when the target contains class labels.

## Real-World Example

A prediction system may have two outcomes:

```text
Pass → 1
Fail → 0
```

The labels are converted into numerical values.

## Business Example

A company may want to predict whether a customer will purchase a product:

```text
Purchased
---------
Yes
No
```

Label Encoding can convert this target into:

```text
Yes → 1
No  → 0
```

## AI/ML Example

In a **customer churn prediction** model:

```text
Churn
-----
Yes
No
```

The target variable can be encoded as:

```text
No  → 0
Yes → 1
```

The ML model can then learn to predict `0` or `1`.

## What does the Python example tell us?

`LabelEncoder` converts the target categories `Yes` and `No` into numerical class labels so that an ML model can work with them.

**Important:** Label Encoding is generally **not preferred for input nominal features**, because assigning numbers such as `0`, `1`, and `2` can incorrectly suggest an order between categories.


````python
## Python Example

from sklearn.preprocessing import LabelEncoder

y = ["Yes", "No", "Yes", "No"]

encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)

print(y_encoded)
````

# Frequency Encoding

## What is it?

Frequency Encoding replaces each category with the **number of times it appears** in the dataset.

**Example:**

```text
City
----
Chennai
Chennai
Bangalore
Mumbai
```

Frequencies:

```text
Chennai   → 2
Bangalore → 1
Mumbai    → 1
```

## Why is it required?

It converts categorical values into numbers while keeping information about **how common each category is**.

## When should we use it?

Use Frequency Encoding when a categorical feature has **many unique categories (high cardinality)** and One-Hot Encoding would create too many columns.

## Real-World Example

A dataset may contain customer locations:

```text
Chennai   → 500 customers
Bangalore → 300 customers
Mumbai    → 200 customers
```

Frequency Encoding represents the cities using these counts.

## Business Example

An e-commerce company may have thousands of different **product categories or brands**.

Instead of creating thousands of One-Hot columns, the company can represent each category based on how frequently it occurs.

## AI/ML Example

In a **fraud detection** model, a `Merchant_ID` feature may contain thousands of unique merchants.

Frequency Encoding can represent:

```text
Merchant_A → 500 transactions
Merchant_B → 20 transactions
Merchant_C → 2 transactions
```

This gives the model information about how frequently each merchant appears.

## What does the Python example tell us?

The `City` column is converted into the **number of times each city appears**. `Chennai` appears 3 times, so every `Chennai` value becomes `3`.

**Important:** Frequency Encoding does not create separate columns for every category, making it useful for **high-cardinality categorical features**.


````python
## Python Example

import pandas as pd

data = pd.DataFrame({
    "City": ["Chennai", "Chennai", "Bangalore", "Mumbai", "Chennai"]
})

frequency = data["City"].value_counts()

data["City_Frequency"] = data["City"].map(frequency)

print(data)
````

# Target Encoding

## What is it?

Target Encoding replaces each category with the **average value of the target variable** for that category.

For example, if we want to predict whether a customer will buy a product:

| City      | Purchase |
| --------- | -------: |
| Chennai   |        1 |
| Chennai   |        1 |
| Chennai   |        0 |
| Bangalore |        0 |
| Bangalore |        0 |
| Mumbai    |        1 |

Target Encoding calculates the average purchase rate for each city:

* Chennai → 2/3 = **0.67**
* Bangalore → 0/2 = **0.00**
* Mumbai → 1/1 = **1.00**

The categorical values are then replaced with these numbers.

## Why is it required?

It converts categorical values into **target-related numerical information**, especially when there are many categories.

## When should we use it?

Use Target Encoding mainly for **high-cardinality categorical features** when the target relationship is important.


## What does the Python example tell us?

It tells us that each city is replaced by its **average purchase rate**.

* Chennai → 0.67
* Bangalore → 0.00
* Mumbai → 1.00

So, the encoded value represents the relationship between the **category and target**.

## Real-World Example

A company wants to predict whether a customer will purchase a product based on their **city**.

Target Encoding can replace each city with its historical purchase rate.

For example:

```text
Chennai   → 0.67
Mumbai    → 0.55
Delhi     → 0.42
```

## Business Example

An e-commerce company has thousands of **product categories** and wants to predict sales.

Instead of creating thousands of One-Hot Encoding columns, Target Encoding can represent each category using its historical average sales or conversion rate.

## AI/ML Use Case

Target Encoding is useful in models such as:

* Customer churn prediction
* Fraud detection
* Credit risk prediction
* Sales prediction
* Recommendation systems
* Conversion prediction

## Important Difference

| Technique           | Uses Target? | Best Use                                             |
| ------------------- | ------------ | ---------------------------------------------------- |
| One-Hot Encoding    | No           | Low-cardinality nominal data                         |
| Ordinal Encoding    | No           | Ordered categories                                   |
| Label Encoding      | No           | Encoding target labels                               |
| Frequency Encoding  | No           | High-cardinality categories                          |
| **Target Encoding** | **Yes**      | High-cardinality categories with target relationship |

**Important:** Target Encoding can cause **data leakage** if the target information is calculated using the same data being used to train the model. In ML, it should be calculated carefully using training data or cross-validation.


````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "City": ["Chennai", "Chennai", "Chennai", "Bangalore", "Bangalore", "Mumbai"],
    "Purchased": [1, 1, 0, 0, 0, 1]
})

# Calculate target mean for each category
target_mean = df.groupby("City")["Purchased"].mean()

# Encode City using target mean
df["City_Target_Encoded"] = df["City"].map(target_mean)

print(df)
````

# Rare Category Grouping

## What is it?

Rare Category Grouping means combining categories that appear **very few times** into a single category such as `"Other"`.

## Why is it required?

It reduces the number of categories and prevents rare categories from creating **unstable or unnecessary features**.

## When should we use it?

Use it when a categorical column contains **many categories with very low frequency**.


## What does the Python example tell us?

It tells us that categories with **fewer than 2 occurrences** are combined into `"Other"`.

* Chennai → kept
* Bangalore → kept
* Mumbai → Other
* Delhi → Other
* Pune → Other

## Real-World Example

A customer dataset may contain thousands of cities:

```text
Chennai       5000
Bangalore     4200
Mumbai        3800
Delhi         3500
Small Town A     2
Small Town B     1
Small Town C     1
```

Rare cities can be grouped:

```text
Small Town A → Other
Small Town B → Other
Small Town C → Other
```

## Business Example

An e-commerce company has customers from many locations. Some locations have only a few customers.

Instead of creating separate features for every small location, the company can group rare locations as `"Other"`.

## AI/ML Use Case

Rare Category Grouping is useful in:

* Customer churn prediction
* Fraud detection
* Sales prediction
* Customer segmentation
* Recommendation systems
* Loan default prediction

## Difference from Other Techniques

| Technique                  | What it does                         | When to use                         |
| -------------------------- | ------------------------------------ | ----------------------------------- |
| One-Hot Encoding           | Creates one column per category      | Low-cardinality categories          |
| Frequency Encoding         | Replaces category with its frequency | High-cardinality categories         |
| Target Encoding            | Replaces category with target mean   | Strong category-target relationship |
| **Rare Category Grouping** | Combines uncommon categories         | Many rare categories                |

**Key idea:** Rare categories → **combine them into `"Other"`**.


````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "City": [
        "Chennai", "Chennai", "Chennai",
        "Bangalore", "Bangalore",
        "Mumbai",
        "Delhi",
        "Pune"
    ]
})

# Count category frequency
frequency = df["City"].value_counts()

# Keep categories appearing at least 2 times
df["City_Grouped"] = df["City"].where(
    df["City"].map(frequency) >= 2,
    "Other"
)

print(df)
````

# Category Cardinality

## What is it?

Category Cardinality means the **number of unique categories** present in a categorical column.

For example:

```text
City
------
Chennai
Mumbai
Delhi
Bangalore
Chennai
```

There are **4 unique categories**, so the cardinality is **4**.

## Why is it required?

It helps us decide **which encoding technique is suitable** for a categorical feature.

## When should we use it?

Check cardinality before encoding, especially when a column contains **many unique categories**.

## What does the Python example tell us?

It tells us that the `City` column contains **4 unique categories**.

```text
Chennai
Mumbai
Delhi
Bangalore
```

So, the cardinality of `City` is **4**.

## Real-World Example

A customer dataset may have:

```text
Gender       → 2 categories
City         → 50 categories
Product      → 10,000 categories
```

Here:

* Gender → Low cardinality
* City → Medium cardinality
* Product → High cardinality

## Business Example

An e-commerce company has a `Product_ID` column containing **100,000 unique products**.

Using One-Hot Encoding would create approximately **100,000 columns**, which is inefficient.

A different technique such as **Frequency Encoding or Target Encoding** may be more appropriate.

## AI/ML Use Case

Cardinality helps select an appropriate encoding technique:

| Cardinality | Example         | Suitable Technique              |
| ----------- | --------------- | ------------------------------- |
| Low         | Gender          | One-Hot Encoding                |
| Low/Medium  | Education Level | Ordinal Encoding                |
| Medium/High | City            | Frequency Encoding              |
| High        | Product ID      | Frequency/Target Encoding       |
| Very High   | Customer ID     | Usually remove or use carefully |

## Difference from Rare Category Grouping

**Category Cardinality** tells us **how many unique categories exist**.

**Rare Category Grouping** reduces the number of categories by combining uncommon ones.

Example:

```text
Before:
100 unique cities

After rare grouping:
20 categories
```

So:

**Cardinality = Number of categories**

**Rare Category Grouping = Reduce unnecessary categories**


````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "City": [
        "Chennai",
        "Mumbai",
        "Delhi",
        "Bangalore",
        "Chennai"
    ]
})

cardinality = df["City"].nunique()

print("Category Cardinality:", cardinality)

````

# Handling Unknown Categories

## What is it?

Unknown Category Handling means dealing with **new categories that were not present in the training data** but appear in new/test data.

For example, during training:

```text
City
------
Chennai
Mumbai
Delhi
```

Later, new data contains:

```text
Bangalore
```

`Bangalore` is an **unknown category**.

## Why is it required?

It prevents errors when the model receives **new categories** that were not seen during training.

## When should we use it?

Use it whenever categorical data may contain **new or unseen categories** during testing or production.

## What does the Python example tell us?

The encoder learned these categories from training data:

```text
Chennai
Mumbai
Delhi
```

When it sees `Bangalore` in test data, it does **not produce an error** because:

```python
handle_unknown="ignore"
```

The unknown category is represented by zeros:

```text
Bangalore → [0, 0, 0]
```

## Real-World Example

A machine learning model is trained with:

```text
Payment Method
----------------
Credit Card
Debit Card
UPI
```

Later, a new payment method such as:

```text
Wallet
```

appears.

The preprocessing system should handle it safely instead of crashing.

## Business Example

An e-commerce company trains a recommendation model using existing product categories.

Later, a new category such as:

```text
Smart Home
```

is introduced.

The model's preprocessing pipeline must handle the new category correctly.

## AI/ML Use Case

Unknown category handling is important in:

* Fraud detection
* Customer churn prediction
* Recommendation systems
* E-commerce prediction
* Credit risk prediction
* Real-time ML applications

## Common Approach

For One-Hot Encoding:

```python
OneHotEncoder(handle_unknown="ignore")
```

This allows unseen categories without producing an error.

## Difference from Other Techniques

| Technique                   | Purpose                                                               |
| --------------------------- | --------------------------------------------------------------------- |
| Rare Category Grouping      | Combines uncommon existing categories                                 |
| Category Cardinality        | Counts unique categories                                              |
| Handling Unknown Categories | Handles categories appearing later that were not seen during training |
| One-Hot Encoding            | Converts categories into binary columns                               |

**Key idea:**

```text
Training data → Chennai, Mumbai, Delhi

New data → Bangalore

Bangalore = Unknown Category
```

The preprocessing pipeline should handle it safely.


````python

## Python Example

from sklearn.preprocessing import OneHotEncoder

X_train = [
    ["Chennai"],
    ["Mumbai"],
    ["Delhi"]
]

X_test = [
    ["Chennai"],
    ["Bangalore"]
]

encoder = OneHotEncoder(handle_unknown="ignore")

encoder.fit(X_train)

X_test_encoded = encoder.transform(X_test)

print(X_test_encoded.toarray())

````

# Combining Categories

## What is it?

Combining Categories means **merging two or more related categories into a single category**.

For example:

```text
Before:
Male
Female
M
F

After:
Male
Female
```

## Why is it required?

It reduces unnecessary categories and makes categorical data **more consistent and useful for ML models**.

## When should we use it?

Use it when different categories represent the **same or similar meaning**.


## What does the Python example tell us?

It tells us that categories representing the same education level are combined.

```text
Bachelors → Bachelor
Masters   → Master
```

This reduces inconsistent category names.

## Real-World Example

A dataset may contain:

```text
USA
US
United States
U.S.
```

These categories represent the same country.

They can be combined into:

```text
United States
```

## Business Example

A company may have customer job categories:

```text
Software Engineer
Developer
Programmer
```

If the business considers all three as the same group, they can be combined:

```text
Software Developer
```

This makes analysis and modeling easier.

## AI/ML Use Case

Combining Categories can be useful in:

* Customer segmentation
* Customer churn prediction
* Fraud detection
* Sales prediction
* Recommendation systems
* Employee analytics

## Difference from Rare Category Grouping

| Technique                  | Purpose                                               |
| -------------------------- | ----------------------------------------------------- |
| **Combining Categories**   | Combines categories because they have similar meaning |
| **Rare Category Grouping** | Combines categories because they occur very few times |
| **One-Hot Encoding**       | Converts categories into separate binary columns      |
| **Target Encoding**        | Converts categories using target information          |
| **Frequency Encoding**     | Converts categories using their frequency             |

### Simple Example

**Combining Categories:**

```text
US + USA + U.S. → United States
```

Reason: **Same meaning**

**Rare Category Grouping:**

```text
Small City A + Small City B + Small City C → Other
```

Reason: **Low frequency**

**Key idea:**

> Combining Categories = Merge similar categories into one meaningful category.


````python
## Python Example

import pandas as pd

df = pd.DataFrame({
    "Education": [
        "Bachelor",
        "Bachelors",
        "Master",
        "Masters",
        "PhD"
    ]
})

# Combine similar categories
df["Education_Combined"] = df["Education"].replace({
    "Bachelors": "Bachelor",
    "Masters": "Master"
})

print(df)
````
