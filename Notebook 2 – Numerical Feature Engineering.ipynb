# Mathematical Transformations

## What is it?

Mathematical transformations create new numerical features by applying mathematical operations to existing features.

## Why is it required?

It helps create more meaningful features that can reveal useful patterns to an ML model.

## When should we use it?

Use it when existing numerical features can be mathematically combined to provide additional useful information.

## What does the Python example tell us?

The code multiplies `price` and `quantity` to create a new feature called `total_amount`.

## AI/ML Example

In a sales prediction model, `price × quantity` can create a **total sales amount** feature.

## Business Example

An e-commerce company can calculate total order value using **price × quantity**.

## AI/ML Engineer Use Case

An AI/ML Engineer creates mathematical features from raw numerical columns to provide additional useful inputs to a machine learning model.


```python

## Python Example

import pandas as pd

data = pd.DataFrame({
    "Price": [100, 200, 150],
    "Quantity": [2, 3, 4]
})

# Create a new feature using multiplication
data["Total_Amount"] = data["Price"] * data["Quantity"]

print(data)
```

```python
## Python Example

price = 100
quantity = 5

total_amount = price * quantity

print(total_amount)
```

# Log Transformation

## What is it?

Log transformation applies a logarithm to a numerical feature to reduce the effect of very large values and skewed data.

## Why is it required?

It reduces skewness and makes highly uneven numerical data easier for a model to learn from.

## When should we use it?

Use it when a numerical feature has a strong right skew and contains positive values.


## What does the Python example tell us?

The code applies a logarithm to the income value, reducing its scale from `100000` to approximately `11.51`.

## AI/ML Example

In a house price prediction model, log transformation can be applied to highly skewed features such as **income or property price**.

## Business Example

A company can use log-transformed **customer spending** to reduce the effect of customers with extremely high spending.

## AI/ML Engineer Use Case

An AI/ML Engineer can apply `np.log1p()` to skewed positive features before training a machine learning model.


```python

## Python Example

import numpy as np

income = 100000

log_income = np.log(income)

print(round(log_income, 2))
```

# Power Transformation

## What is it?

Power transformation changes a numerical feature using a mathematical power to make its distribution more suitable for machine learning.

## Why is it required?

It helps reduce skewness and make numerical data easier for models to learn from.

## When should we use it?

Use it when a numerical feature is significantly skewed or has a non-normal distribution.


## What does the Python example tell us?

The code raises `4` to the power of `2`, creating a transformed value of `16`.

## AI/ML Example

A model can use power transformations to make skewed features such as **income or transaction amounts** more suitable for learning.

## Business Example

A bank can transform highly skewed **transaction values** before using them for customer analysis.

## AI/ML Engineer Use Case

An AI/ML Engineer can use techniques such as **Yeo-Johnson** or **Box-Cox** power transformations to reduce feature skewness before model training.


```python

## Python Example

import numpy as np

value = 4

squared_value = np.power(value, 2)

print(squared_value)
```

# Absolute Difference

## What is it?

Absolute difference calculates the size of the difference between two numerical values without considering whether the difference is positive or negative.

## Why is it required?

It focuses only on how far apart two values are.

## When should we use it?

Use it when the direction of the difference does not matter.

## What does the Python example tell us?

The code calculates the difference between the actual and expected price and uses `abs()` to return the positive gap of `200`.

## AI/ML Example

A recommendation model can use the absolute difference between **customer age and product target age** to measure similarity.

## Business Example

A company can calculate the absolute difference between **forecasted sales and actual sales** to measure prediction error.

## AI/ML Engineer Use Case

An AI/ML Engineer can create absolute-difference features when comparing two related numerical features where only the distance between them matters.


```python

## Python Example

actual_price = 800
expected_price = 1000

absolute_difference = abs(actual_price - expected_price)

print(absolute_difference)

```

# Relative Difference

## What is it?

Relative difference measures the difference between two values relative to their average or reference value.

## Why is it required?

It helps compare differences fairly when the values have different scales.

## When should we use it?

Use it when the percentage or proportional gap between two values is more useful than the absolute gap.


## What does the Python example tell us?

The code shows that actual sales are **10% lower** than expected sales.

## AI/ML Example

A forecasting model can use the relative difference between **predicted sales and actual sales** as a feature.

## Business Example

A company can measure how much **actual revenue differs from the target revenue** in percentage terms.

## AI/ML Engineer Use Case

An AI/ML Engineer can create relative-difference features to capture proportional changes between predicted, actual, or reference values.


```python

## Python Example

actual_sales = 900
expected_sales = 1000

relative_difference = abs(actual_sales - expected_sales) / expected_sales

print(relative_difference)
```

# Normalized Features

## What is it?

Normalized features are numerical features transformed to a common scale.

## Why is it required?

It prevents features with larger values from dominating features with smaller values.

## When should we use it?

Use it when numerical features have very different ranges and the ML algorithm is sensitive to feature scale.

## What does the Python example tell us?

The code converts the values from `10–30` to a common range of `0–1`.

## AI/ML Example

A machine learning model can normalize features such as **age, income, and account balance** before training.

## Business Example

A company can normalize **sales, profit, and customer spending** so they can be compared on the same scale.

## AI/ML Engineer Use Case

An AI/ML Engineer can use `MinMaxScaler` to normalize numerical features before training models such as **KNN, SVM, and neural networks**.


```python

## Python Example

from sklearn.preprocessing import MinMaxScaler

data = [[10], [20], [30]]

scaler = MinMaxScaler()
normalized_data = scaler.fit_transform(data)

print(normalized_data.flatten())
```
